# Importando bibliotecas

```python
import pandas as pd
```

```python

# Importando dados
df_clientes = pd.read_csv('../data/raw/customers.csv')
df_products = pd.read_csv('../data/raw/products.csv')
df_stores = pd.read_csv('../data/raw/stores.csv')
df_sales = pd.read_csv('../data/raw/sales.csv')
df_items_sales = pd.read_csv('../data/raw/items_sales.csv')
```

```python
# Unindo a tabela de produtos com a de items de vendas
df_items_sales = pd.merge(df_items_sales, df_products, on='id_product', how='inner')
df_items_sales.head()
```

# Qual categoria de produtos teve mais lucro?

```python
# Primeiro vamos criar uma nova coluna, multiplicando a quantidade pelo preco
df_items_sales['total'] = df_items_sales['quantity'] * df_items_sales['price']
df_items_sales.head()
```

```python
category_money = (
    df_items_sales.groupby('category')['total']
    .sum()
    .reset_index(name='total_vendas')
    .sort_values(by='total_vendas', ascending=False)
)
```

```python
category_money
```

Nosso maior lucro vem da categoria de games, seguido por áudio e Cadeiras.

```python
# total de vendas

total_sales = category_money['total_vendas'].sum()
print(f"Nosso faturamento total foi de R$ {total_sales}")

```

```python
# Vamos descobrir como as categorias representam em porcentagem para o valor final.
category_money['porcentagem'] = (category_money['total_vendas'] / total_sales) * 100
category_money['porcentagem'] = round(category_money['porcentagem'],2)
category_money
```

Vamos agora descobrir o quanto gastamos para posteriormente descobrir o lucro líquido.

```python
df_items_sales['custo_total'] = df_items_sales['quantity'] * df_items_sales['custo']
df_items_sales.head()
```

```python
# Custo por category

custo_category_money = (
    df_items_sales.groupby('category')['custo_total']
    .sum()
    .reset_index(name='custo_total_compras')
    .sort_values(by='custo_total_compras', ascending=False)
)
custo_category_money.head()
```

```python
total_custo = custo_category_money['custo_total_compras'].sum()
```

```python
faturamento_liquido_vendas = total_sales - total_custo
print(f'Nosso faturamento líquido é de R$ {faturamento_liquido_vendas}')
```
